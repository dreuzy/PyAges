## Calibration
### Parameters
Parameters for the Forward error propagation and Metropolis Hastings Methods 

```python
# ---------------- CALIBRATION PARAMETERS ----------------
run_calibration_simplex or run_calibration_metropolis_hastings_metropolis_hastings = True
run_calibration_simplex or run_calibration_metropolis_hastings_simplex = True
calibration_strategies = [None] * 2

# ---------------- FORWARD UNCERTAINTY QUANTIFICATION -----------------------------
calibration_strategies[0] = csimp.Simplex("forward_uncertainty_quantification",
                                            init_multiples_n=5,fuq_n=50)

# ---------------- METROPOLIS HASTINGS --------------------
# Method and Parameters  
calibration_strategies[1] = cMH.MetropolisHastings(nstep=25000,prior_option=False,likelihood=True,
                                                     monitor=True,display_traj=True)
# calibration_strategies[1].MH_step.define_by_prop(0.005)
calibration_strategies[1].MH_step.define_by_value()
```

### Calibration per se

```python
lpm_results=[None]*2
# ---------------- CALIBRATION -------------
for i in range(len(calibration_strategies)):
    # Outputs of Interpration
    directory_calibration = gp.results_directory(display.directory,calibration_strategies[i].method)
    display.directory = directory_calibration
    # Calibration
    calib_basis=calbas.CalibrationCore(concentration_sampled, lpm_model_name, display_options=display, lpm_data_directory=lpm_data_directory)
    calib_basis.prepare()
    calibration_strategies[i].update_calibbasis(calib_basis)
    lpm_results[i]=calibration_strategies[i].perform()
    # Stores/Writes Results
    calibration_strategies[i].write_calibrated_lpm(lpm_results[i])
```
